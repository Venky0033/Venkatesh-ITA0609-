```python
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------
# SALES DATA - 15 EXAMPLES
# -------------------------------

products = np.array([
    "Laptop", "Phone", "Tablet", "Headphones", "Smartwatch",
    "Laptop", "Phone", "Camera", "Tablet", "Monitor",
    "Headphones", "Keyboard", "Mouse", "Camera", "Phone"
])

sales = np.array([
    75000, 45000, 30000, 8000, 15000,
    75000, 45000, 55000, 30000, 25000,
    8000, 5000, 3000, 55000, 45000
])

quantity = np.array([
    3, 5, 4, 10, 6,
    3, 5, 4, 4, 7,
    10, 12, 15, 4, 5
])

rating = np.array([
    4.5, 4.2, 4.0, 4.3, 4.1,
    4.5, 4.2, 4.4, 4.0, 4.3,
    4.3, 4.1, 4.0, 4.4, 4.2
])

print("SALES DATA")
print(products)
print(sales)
print(quantity)
print(rating)


# -------------------------------
# 1. FILTERING DATA
# -------------------------------

print("\n1. Sales greater than 30000:")
print(sales[sales > 30000])


# -------------------------------
# 2. MULTIPLE CONDITIONS
# -------------------------------

print("\n2. Sales > 30000 and Rating > 4.2:")

condition = (sales > 30000) & (rating > 4.2)

print(products[condition])
print(sales[condition])


# -------------------------------
# 3. SORTING ASCENDING
# -------------------------------

print("\n3. Sales Ascending:")
print(np.sort(sales))


# -------------------------------
# 4. SORTING DESCENDING
# -------------------------------

print("\n4. Sales Descending:")
print(np.sort(sales)[::-1])


# -------------------------------
# 5. REMOVING DUPLICATES
# -------------------------------

print("\n5. Unique Sales:")
print(np.unique(sales))


# -------------------------------
# 6. AGGREGATION / MATH FUNCTIONS
# -------------------------------

print("\n6. Math Functions")

print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Maximum Sales:", np.max(sales))
print("Minimum Sales:", np.min(sales))
print("Median Sales:", np.median(sales))
print("Standard Deviation:", np.std(sales))
print("Variance:", np.var(sales))

print("Square Root:")
print(np.sqrt(sales))

print("Square:")
print(np.square(sales))


# -------------------------------
# 7. LINE PLOT
# -------------------------------

plt.figure(figsize=(10, 5))

plt.plot(products, sales, marker="o")

plt.title("Product vs Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# -------------------------------
# 8. BAR CHART
# -------------------------------

plt.figure(figsize=(10, 5))

plt.bar(products, sales)

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# -------------------------------
# 9. PIE CHART
# -------------------------------

plt.figure(figsize=(8, 8))

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%"
)

plt.title("Sales Distribution")

plt.show()


# -------------------------------
# 10. HISTOGRAM
# -------------------------------

plt.figure(figsize=(8, 5))

plt.hist(sales, bins=5)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()


# -------------------------------
# 11. SCATTER PLOT
# -------------------------------

plt.figure(figsize=(8, 5))

plt.scatter(quantity, sales)

plt.title("Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")

plt.show()
```
